```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("zomato.csv", encoding="latin-1")
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

```

```python
delivery_count = df['Has Online delivery'].value_counts()
print("Number of restaurants with and without online delivery:")
print(delivery_count)
```

```python
delivery_percentage = df['Has Online delivery'].value_counts(normalize=True) * 100
print("Percentage of restaurants:")
print(delivery_percentage.round(2))
```

```python
plt.figure(figsize=(7,5))
delivery_percentage.plot(kind='bar')
plt.title('Percentage of Restaurants Providing Online')
plt.xlabel('Online Delivery')
plt.ylabel('Percentage(%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
delivery_rating = df.groupby('Has Online delivery')['Aggregate rating'].mean()
print("Average rating by Online Delivery:")
print(delivery_rating.round(2))
```

```python
plt.figure(figsize=(7,5))
delivery_percentage.plot(kind='bar')
plt.title('Average Rating: Online Delivery vs No Online Delivery')
plt.xlabel('Online Delivery')
plt.ylabel('Average Rating')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
print("========== ONLINE DELIVERY ANALYSIS ==========")
print("\n1.NUMBER OF RESTAURANTS:")
print(delivery_count)
print("\n2.PERCENTAGE OF RESTAURANTS:")
print(delivery_percentage.round(2))
print("\n3.AVERAGE RATING:")
print(delivery_rating.round(2))
```

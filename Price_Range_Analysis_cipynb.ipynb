```python
import pandas as pd
import matplotlib.pyplot as plt
df= pd.read_csv("zomato.csv", encoding="latin-1")
print("Dataset loaded successfully!")
print(df.shape)
```

```python
price_count = df['Price range'].value_counts().sort_index()
print("Number of Restaurants in Each Price Range:")
print(price_count)
```

```python
plt.figure(figsize=(8, 5))
price_count.plot(kind='bar')
plt.title('Distribution of Restaurants by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
plt.figure(figsize=(7,7))
price_count.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
    )
plt.title('Percentage of Restaurants by Price Range')
plt.ylabel('')
plt.show()
```

```python
price_rating=df.groupby('Price range')['Aggregate rating'].mean()
print("Average Rating for Price Range:")
print(price_rating.round(2))
```

```python
highest_rating_range=price_rating.idxmax()
highest_rating=price_rating.max()
print("Price range with Highest Average Rating:",highest_rating_range)
print("Highest Average Rating:",round(highest_rating, 2))
```

```python
plt.figure(figsize=(8, 5))
price_rating.plot(kind='bar')
plt.title('Average Restaurant Rating by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Average Rating')

plt.tight_layout()
plt.show()
#
```
